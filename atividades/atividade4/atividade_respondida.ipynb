```python
import pandas as pd

df = pd.read_csv("estoque_supermercado.csv")
print(df.info)
```

LIMPEZA DO PREÇO

```python
#
df["preco"] = (
    df["preco"]
    .astype("string")
    .str.replace("R$","",regex=False)
    .str.replace(" ","",regex=False)
    .str.replace(",",".",regex=False)
    .str.strip()
)

df["preco"] = pd.to_numeric(df["preco"],
    errors="coerce")
print(df["preco"].head(10))
print(df["preco"].dtype)

```
